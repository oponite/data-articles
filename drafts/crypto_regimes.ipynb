# BTC Accumulation Strategy Backtest Study

## Purpose

This study tested several BTC accumulation strategies over a 5-year period to compare simple recurring buying, adaptive value averaging, and buy-the-dip rules.

The goal was not just to see which strategy made the most money, but also to understand how much capital each strategy required, how efficiently it used that capital, and how painful the drawdowns were.

## Backtest Setup

- Asset: BTC-USD
- Period: July 3, 2021 through July 2, 2026
- Length: 5 years
- Granularity: daily closes
- Data source: Coinbase daily candles
- Missing days: 0
- Fees/slippage/taxes: not included
- Execution assumption: buys happen at the daily close
- Strategy type: long-only accumulation
- No selling, except no strategy here sells BTC
- Walk-forward testing: 30-day training window, followed by 30-day test windows

## Strategies Tested

### 1. Dollar-Cost Averaging

Dollar-cost averaging means buying a fixed dollar amount on a fixed schedule, regardless of price.

Variants tested:

- Buy $1 of BTC every day
- Buy $10 of BTC every week
- Buy $50 of BTC every month

### 2. Value Averaging

Value averaging does not buy a fixed amount each period. Instead, it tries to keep the portfolio on a planned value path.

Example: if the target path is $50/month, then after 12 months the BTC position target is about $600. If the BTC position is worth $520, the strategy buys $80. If it is already worth $700, it buys nothing.

Variants tested:

- $1/day target path
- $10/week target path
- $50/month target path

### 3. Buy-the-Dip

Buy-the-dip strategies only buy after BTC has already dropped by a certain amount.

Variants tested:

- Buy $1 after BTC falls at least 5% in 1 day
- Buy $10 after BTC falls at least 10% in 1 week
- Buy $50 after BTC falls at least 20% in 1 month

## Metrics Explained

- Total invested: total dollars contributed into the strategy.
- Ending value: final BTC position value at the end of the backtest.
- Profit: ending value minus total invested.
- Return on contributed cash: profit relative to the money actually invested.
- Money-weighted IRR: annualized return that accounts for when cash was invested.
- Max drawdown: worst peak-to-trough decline in account value.
- Days below contributed capital: percent of days where the account was worth less than total money invested.

## Walk-Forward Results

The walk-forward test split the backtest into repeated 30-day test windows. This checks whether the strategies performed consistently across time instead of only looking good over the full 5-year period.

| Strategy | Avg 30-Day Test Return | Median 30-Day Test Return | Positive Test Folds |
|---|---:|---:|---:|
| DCA / value averaging variants | 1.92% | 1.35% | 55.0% |
| Buy dip -10% in 1 week | 1.68% | 1.01% | 53.3% |
| Buy dip -5% in 1 day | 1.65% | 1.01% | 53.3% |

## Interpretation

The buy-the-dip strategy using a 20% monthly drop had the highest ending value. It turned $6,250 into $11,349. However, it also invested the most cash by far. Its strong result came partly from deploying a lot of money during large BTC drawdowns.

The value averaging strategies were the most capital-efficient. The $50/month value averaging path invested only about $1,387 and ended at about $3,332. That means it made more profit than monthly DCA while using less than half the cash.

DCA was simple and solid. The daily, weekly, and monthly DCA versions all produced similar percentage returns, around 46% to 47%. The main difference was how much money each version invested.

The smaller buy-the-dip rules were less useful in dollar terms. The -5% daily dip rule had a decent percentage return, but it only invested $67 total, so the final account value was tiny.

Lump-sum BTC had a good final return but the worst drawdown. It fell about 77% from peak to trough, which would have been psychologically difficult to hold through.

## Conclusion

The best strategy depends on the goal.

If the goal is the highest ending account value, the -20% monthly dip-buying strategy performed best because it deployed the most money during major BTC weakness.

If the goal is capital efficiency, value averaging was the strongest. It used much less cash than DCA while producing better return on contributed capital and higher IRR.

If the goal is simplicity, DCA is still very reasonable. It was easy to implement, required no price judgment, and produced solid results across daily, weekly, and monthly schedules.

The most practical takeaway is that value averaging deserves further testing. It combines discipline with adaptability: it buys when BTC is behind the target path and skips when BTC is already ahead. For a personal accumulation strategy, that can preserve cash while still building a BTC position over time.

A reasonable next version would be a monthly BTC value averaging strategy with:
- a $100/month target growth path,
- a maximum buy cap,
- no selling,
- duplicate-buy protection,
- and a dry-run mode before connecting to a real Coinbase account.
